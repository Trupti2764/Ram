```python
import numpy as np

# Define two fuzzy sets
A = np.array([0.2, 0.5, 0.7, 1.0])
B = np.array([0.6, 0.4, 0.8, 0.3])

# Define fuzzy relation matrices
R = np.array([[0.2, 0.7],
              [0.5, 0.9]])

S = np.array([[0.6, 0.3],
              [0.8, 0.4]])


# Function for Union
def fuzzy_union():
    result = np.maximum(A, B)
    print("\nFuzzy Set A:", A)
    print("Fuzzy Set B:", B)
    print("Union (A U B):", result)


# Function for Intersection
def fuzzy_intersection():
    result = np.minimum(A, B)
    print("\nFuzzy Set A:", A)
    print("Fuzzy Set B:", B)
    print("Intersection (A ∩ B):", result)


# Function for Complement
def fuzzy_complement():
    result = 1 - A
    print("\nFuzzy Set A:", A)
    print("Complement of A:", result)


# Function for Max-Min Composition
def max_min_composition():
    T = np.zeros((R.shape[0], S.shape[1]))

    for i in range(R.shape[0]):
        for j in range(S.shape[1]):
            T[i, j] = np.max(np.minimum(R[i, :], S[:, j]))

    print("\nRelation Matrix R:\n", R)
    print("\nRelation Matrix S:\n", S)
    print("\nMax-Min Composition (R o S):\n", T)


# Main Menu
while True:

    print("\n========== FUZZY LOGIC OPERATIONS ==========")
    print("1. Fuzzy Union")
    print("2. Fuzzy Intersection")
    print("3. Fuzzy Complement")
    print("4. Max-Min Composition")
    print("5. Display All Operations")
    print("6. Exit")

    choice = input("\nEnter your choice (1-6): ")

    if choice == '1':
        fuzzy_union()

    elif choice == '2':
        fuzzy_intersection()

    elif choice == '3':
        fuzzy_complement()

    elif choice == '4':
        max_min_composition()

    elif choice == '5':
        fuzzy_union()
        fuzzy_intersection()
        fuzzy_complement()
        max_min_composition()

    elif choice == '6':
        print("\nExiting Fuzzy Logic Program. Thank you!")
        break

    else:
        print("\nInvalid choice! Please enter a number between 1 and 6.")
```
